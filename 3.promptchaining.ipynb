```python
from langgraph.graph import StateGraph,START,END
from langchain_openai import ChatOpenAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model=ChatOpenAI(model="gpt-5.6-luna")
```

```python
class LLMState(TypedDict):
    title:str
    outline:str
    content:str
```

```python
def create_outline(state:LLMState)->LLMState:
    title=state['title']

    prompt=f"Generate a detailed outline for a blog on the topic - {title}"
    outline=model.invoke(prompt)
    state['outline']=outline
    return state
```

```python
def create_blog(state:LLMState)->LLMState:
    title=state['title']
    outline=state['outline']
    prompt=f"Generate a blog using this title : {title} and outline : {outline}"
    blog=model.invoke(prompt).content
    state['content']=blog
    return state
```

```python
graph=StateGraph(LLMState)

graph.add_node("create_outline",create_outline)
graph.add_node("create_blog",create_blog)

graph.add_edge(START,"create_outline")
graph.add_edge("create_outline","create_blog")
graph.add_edge("create_blog",END)

workflow=graph.compile()
```

```python
initial_state={"title":"AGENTIC AI BY 2050"}
final_state=workflow.invoke(initial_state)
print(final_state)
```
